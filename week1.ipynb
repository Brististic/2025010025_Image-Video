```python
from google.colab.patches import cv2_imshow
from IPython.display import Video
```

```python
image=cv2.imread("cattos.jpg")
print("Original Image")
cv2_imshow(image)
```

```python
cv2.imwrite("savedcattos.jpg",image)
print("Image saved.")
```

```python
cv2.imwrite("cattos.png", image)
cv2.imwrite("cattos.tiff", image)
cv2.imwrite("cattos.bmp", image)

print("Image converted to PNG, TIFF, and BMP formats.")
```

```python
resized = cv2.resize(image, (400, 300))

print("Resized Image")
cv2_imshow(resized)


cv2.imwrite("resized_image.jpg", resized)

print("Image resized successfully.")
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print("Grayscale Image")
cv2_imshow(gray)

cv2.imwrite("gray_image.jpg", gray)

print("Grayscale image saved.")
```

```python
print("Displaying Video")

Video("video.mp4", embed=True)
```

```python
cap = cv2.VideoCapture("video.mp4")

count = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    filename = f"frame_{count}.jpg"

    cv2.imwrite(filename, frame)

    count += 1

cap.release()

print("Total Frames Extracted:", count)
```
