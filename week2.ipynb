```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
from google.colab import files
uploaded=files.upload()
filename = list(uploaded.keys())[0]
```

```python
gray_img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)
```

```python
mean_threshold = np.mean(gray_img)
binary_case1 = np.where(gray_img >= mean_threshold, 255, 0).astype(np.uint8)
```

```python
user_threshold = float(input("\nEnter threshold value (0 to 255): "))
binary_case2 = np.where(gray_img >= user_threshold, 255, 0).astype(np.uint8)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(gray_img, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(binary_case1, cmap='gray')
plt.title(f'Binary (Mean T={mean_threshold:.1f})')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(binary_case2, cmap='gray')
plt.title(f'Binary (User T={user_threshold})')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
print("Please upload a color photo:")
uploaded = files.upload()
filename = list(uploaded.keys())[0]
```

```python
bgr_img = cv2.imread(filename)
rgb_img = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2RGB)
```

```python
R = rgb_img[:, :, 0].astype(np.float32)
G = rgb_img[:, :, 1].astype(np.float32)
B = rgb_img[:, :, 2].astype(np.float32)
```

```python
gray_case1 = ((R + G + B) / 3.0).astype(np.uint8)
```

```python
w_r = 0.7
w_g = 0.2
w_b = 0.1
```

```python
gray_case2 = (w_r * R + w_g * G + w_b * B).astype(np.uint8)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(rgb_img)
plt.title('Original RGB')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(gray_case1, cmap='gray')
plt.title('Grayscale (Mean Avg)')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(gray_case2, cmap='gray')
plt.title(f'Grayscale ({w_r}, {w_g}, {w_b})')
plt.axis('off')

plt.tight_layout()
plt.show()
```
